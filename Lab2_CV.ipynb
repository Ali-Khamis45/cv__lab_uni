###Image Manipulation

manipulation on the image is done using arithmetic operations, logic operations, and accessing array elements and sub-array.

to increase brightness in the image.


```python
BrightnessIncrease = image + 50
ski.io.imshow(BrightnessIncrease)
```

This increase image brightness by 50.

To increase a certain color in the image.


```python
BrightnessIncrease = image.copy()
BrightnessIncrease[:,:,0] += 50
ski.io.imshow(BrightnessIncrease)
```

First, we copy the image to a new variable, because we don't want to change the original image when increasing brightness of the new variable (reference type). Then you access the red channel and increase intensity of it, we didn't write it as follows BrightnessIncrease = BrightnessIncrease[:,:,0] + 50, because BrightnessIncrease will be re-assigned with one color channel which make the image gray not colored.

### To add lines in the image

```python
image = NatrureImage.copy()

#red line
image[0:100] = (255,0,0)
ski.io.imshow(image)
```

```python
image[:,:200] = (0,255,0)
ski.io.imshow(image)
```

###Draw plus sign

Draw a plus sign in the middle of a gray image.

```python
imCamera = ski.data.camera()
height = imCamera.shape[0]
midHeight = int(imCamera.shape[0]/2)
width = imCamera.shape[1]
midWidth = int(imCamera.shape[1]/2)
ski.io.imshow(imCamera)
```

```python
imCamera[midHeight-50:midHeight+50] = 0
ski.io.imshow(imCamera)
```

```python
imCamera[:,midWidth-50:midWidth+50] = 0
ski.io.imshow(imCamera)
```

###Split image into multiple colors

```python
red = NatrureImage.copy()
green = NatrureImage.copy()
blue = NatrureImage.copy()

red[:,:,1:3] = 0

green[:,:,0] = 0
green[:,:,2] = 0

blue[:,:,0:2] = 0
```

```python
ski.io.imshow(red)
```

```python
ski.io.imshow(green)
```

```python
ski.io.imshow(blue)
```

###Combine images from multiple colors.

After we split the image into 3 channels we can create a new 3D black image. then add red, green, and blue images together to get the final image. we can add red with green and blue because red color is only presented in the 0-index in the 3rd dimension `[:,:,0]` and green present in the 1-index in the 3rd dimension `[:,:,1]` and blue present in the 2-index in the 3rd dimension `[:,:,2]`.  

```python
n = np.zeros((NatrureImage.shape[0],NatrureImage.shape[1],NatrureImage.shape[2]), dtype=np.uint8)
n = red + green + blue
ski.io.imshow(n)
```
