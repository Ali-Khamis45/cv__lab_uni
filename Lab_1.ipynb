##Definitions

Before starting, it is crucial to understand how Python handles memory and variables.

 **Value Type vs. Reference Type**


*   Value Type: The data stored in the variable is copied, not its address. Changes to the new variable do not affect the original.
*   Reference Type: The address of the data is copied. Changes to the structure of the new variable will affect the original.

```python

x = 10
y = x
y = 20
print(f"X: {x}, Y: {y}")

# Output: X: 10, Y: 20
# value of x (10) was copied. Changing y did not affect x.


List_1 = [1, 2, 3]
List_2 = List_1         # Copies the address, not the value
List_2.append(5)

print(f"List 1: {List_1}, List 2: {List_2}")


# Output: List 1: [1,2,3,5], List 2: [1,2,3,5]
```

##Mutable vs. Immutable



*   Immutable: Once the object is created, its content cannot be changed. If content is modified, a new memory address is allocated.
*   Mutable: The content of the variable can be changed without changing the memory address.



```python
# --- Immutable Example (int) ---
x = 1
print(f"x is {x} and x address : {id(x)}")
y = x
print(f"y is {y} and y address : {id(y)}")

x = 2 # Modifying x allocates a new address

print(f"x is {x} and x address : {id(x)}")
print(f"y is {y} and y address : {id(y)}")
# Note: x's address changed.
```

```python
# --- Mutable Example (list) ---
x = [1, 2, 3, 4, 5]
print(f"x is {x} and x address : {id(x)}")

y = x
print(f"y is {y} and y address : {id(y)}")

x[0] = 100 # Modifying content matches the definition of Mutable
print(f"x is {x} and x address : {id(x)}")
print(f"y is {y} and y address : {id(y)}")
# Note: Even after modifying x, the address remains the same.
```

## Python Data Types

Python has 10 basic data types including int, float, string, Boolean, list, set, tuple, dictionary, none, and objects.

### Integers (int)


*   Stores whole numbers.



```python
x = 5
# OR
x = int() # Performance drawback compared to direct assignment

# Demonstration of Immutability
x = 5
print(f"Address of 5: {id(x)}")
x = 7
print(f"Address of 7: {id(x)}") # Address changes
```

### Float (float)
*Stores real numbers.*

```python
x = 5.3
# OR
x = float()
# Demonstration of Value Type
x = 5.0
y = x
x = 7.0
print(f"Y remains: {y}") # Prints 5.0
```

### Boolean (bool)
Stores True or False.

```python
x = True
y = x
x = False
print(f"Y is: {y}") # Prints True
```

### String (str)
Holds words/text.

```python
x = "Hello"
# OR
x = str()
```

### List (list)

```python
x = [1, "hello"]

# Reference behavior
x = [1, 2, 3, 4, 5]
y = x
x[0] = 100
print(f"Y reflects change: {y}") # [100, 2, 3, 4, 5]
```

##Other Structures

Set: Collection of unique data. Reference and Mutable.

```python
x = {1, 2, 3}
```

Tuple: Holds data in a pair manner. Reference and Immutable.

```python
x = (1, 2, 3)
```

Dictionary: Key-value pairs. Reference and Mutable.

```python
x = {'a' : 1}
```

##Control Flow

###Conditions

Branches activated based on boolean truth

```python
# If example
if True:
    print("Hello world 1")

# Else example
if False:
    print("Hello world 1")
else:
    print("Hello world 2") # This runs

# Elif example
if False:
    print("Hello world 1")
elif True:
    print("Hello world 2") # This runs
```

### Loops

Blocks of statements that run multiple times.

```python
for i in range(5):
    print(i, end="-")
# Output: 0-1-2-3-4
```

##Functions

Blocks of code that can be called at any time.

```python
def greet():
    print("Hello, world!")

greet()
# Output: Hello, world!
```
